```python
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np

# Step 1: List your image filenames
image_paths = [
    "E1 spectrogram.jpg",
    "E2 spectrogram.jpg",
    "E3 spectrogram.jpg",
    "E4 spectrogram.jpg",
    "E5 spectrogram.jpg",
    "E1 spectrum.jpg",
    "E2 spectrum.jpg",
    "E3 spectrum.jpg",
    "E4 spectrum.jpg",
    "E5 spectrum.jpg"
]

# Step 2: Load and resize images to same size (200x200)
images = [Image.open(path).resize((200, 200)) for path in image_paths]
images_np = [np.array(img) for img in images]

# Step 3: Create the 5x2 grid plot
fig, axs = plt.subplots(5, 2, figsize=(6, 10))  # Adjust figsize if needed

for ax, img in zip(axs.flat, images_np):
    ax.imshow(img)
    ax.axis('off')

plt.tight_layout()

# Step 4: Save the figure as an image
# plt.savefig("grid.jpg", bbox_inches='tight', dpi=300)
plt.savefig("plot.pdf", dpi=300, bbox_inches='tight')
plt.close()

print("Image grid saved as grid_output.jpg")
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np

# Step 1: List of image filenames (replace with your actual filenames)
image_paths = [
    "E1 spectrogram.jpg",
    "E1 spectrum.jpg",
    "E2 spectrogram.jpg",
    "E2 spectrum.jpg",
    "E3 spectrogram.jpg",
    "E3 spectrum.jpg",
    "E4 spectrogram.jpg",
    "E4 spectrum.jpg",
    "E5 spectrogram.jpg",
    "E5 spectrum.jpg"
]

# Step 1: Load and resize all images to 800x500
target_width, target_height = 800, 500
images = [Image.open(path).resize((target_width, target_height)) for path in image_paths]
images_np = [np.array(img) for img in images]

# Step 2: Manually create a blank canvas to paste images — avoids ALL subplot spacing issues
rows, cols = 5, 2
grid_width = cols * target_width     # 1600
grid_height = rows * target_height   # 2500

# Create blank canvas
grid_image = Image.new("RGB", (grid_width, grid_height))

# Step 3: Paste each image at the correct (x, y) position
for idx, img in enumerate(images):
    row = idx // cols
    col = idx % cols
    x = col * target_width
    y = row * target_height
    grid_image.paste(img, (x, y))

# Step 4: Save the final image
grid_image.save("GRID.pdf")
print("Saved: tight_image_grid.pdf")
```
